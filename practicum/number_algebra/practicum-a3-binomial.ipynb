## Теория 1. Формула и треугольник

Для целого положительного $n$

$$(a+b)^n = \sum_{r=0}^{n} \binom{n}{r} a^{\,n-r} b^{\,r}
= a^n + \binom{n}{1}a^{n-1}b + \binom{n}{2}a^{n-2}b^2 + \dots + b^n,$$

где $\displaystyle\binom{n}{r} = \frac{n!}{r!\,(n-r)!}$.

Два свойства, которые экономят время:

**Симметрия.** $\binom{n}{r} = \binom{n}{n-r}$. Считать $\binom{9}{7}$ незачем —
это $\binom{9}{2} = 36$. На экзамене это же значит, что нужный член можно
искать с любого конца: $r = 2$ и $r = 6$ в $(ax^3+b)^8$ дают один
и тот же коэффициент.

**Треугольник Паскаля.** Каждая строка получается сложением соседей
предыдущей:

```
n=0                1
n=1              1   1
n=2            1   2   1
n=3          1   3   3   1
n=4        1   4   6   4   1
n=5      1   5  10  10   5   1
n=6    1   6  15  20  15   6   1
n=7  1   7  21  35  35  21   7   1
```

До $n = 7$ строку быстрее дописать, чем считать факториалы, и в Paper 1
это законный способ: markscheme прямо принимает «lists terms from row 7
of Pascal's triangle». Выше $n = 7$ — формула.

**Где теряют балл.** В $(a+b)^n$ буквой $a$ может оказаться что угодно.
В $(3+x^2)^8$ первое слагаемое равно 3, и его степень $3^{\,8-r}$ никуда
не девается; в $(x^3 - \frac{1}{8x})^n$ второе слагаемое отрицательно,
и знак $(-1)^r$ обязан появиться в ответе.

## Теория 2. Общий член: не раскрывать, а искать номер

Ключевой приём темы, 27 баллов архива напрямую и ещё 50 внутри других
приёмов. Член разложения $(a+b)^n$ с номером $r+1$ равен

$$T_{r+1} = \binom{n}{r}\,a^{\,n-r} b^{\,r}, \qquad r = 0, 1, \dots, n.$$

Порядок работы всегда один:

1. Выписать общий член, подставив вместо $a$ и $b$ то, что стоит в скобке.
2. Собрать все степени $x$ в одну: показатель окажется **линейной функцией
   от $r$**.
3. Приравнять этот показатель к нужному и найти $r$.
4. Подставить $r$ обратно и посчитать коэффициент.

Пример на $\left(x^3 - \dfrac{1}{8x}\right)^n$. Общий член:

$$\binom{n}{r}\left(x^3\right)^{n-r}\left(-\frac{1}{8x}\right)^{r}
= \binom{n}{r}\left(-\frac18\right)^{r} x^{3(n-r)} x^{-r}
= \binom{n}{r}\left(-\frac18\right)^{r} x^{\,3n-4r}.$$

Показатель $3n - 4r$ — вся суть задачи. Нужен постоянный член? Значит
$3n - 4r = 0$. Нужен $x^6$? Значит $3n - 4r = 6$.

**Три места, где теряют баллы.**

*Нумерация.* Член с номером $r+1$, а не $r$. «Третий член» — это $r = 2$.
Путаница стоит балла и обычно всей задачи.

*Член против коэффициента.* «The coefficient of $x^5$» — это число перед
$x^5$, без самого $x^5$. «The term in $x^5$» — со степенью. Читайте, что
просят.

*Дробное $r$.* Если из уравнения выходит $r = 5/3$, члена с такой степенью
**нет**, и это часто и есть ответ («не существует» или «только при таких $n$»).
Округлять $r$ нельзя.

## Теория 3. Приравнять коэффициенты

Если разложение уже написано в условии, а в скобке стоят буквы, работа
идёт в обратную сторону: коэффициенты, посчитанные формулой, приравнивают
к данным и решают систему.

Схема на $(1+kx)^n$:

$$(1+kx)^n = 1 + \binom{n}{1}kx + \binom{n}{2}k^2x^2 + \dots
= 1 + nkx + \frac{n(n-1)}{2}k^2x^2 + \dots$$

Дано $1 + 12x + 28k^2x^2$. Приравниваем:

$$nk = 12, \qquad \frac{n(n-1)}{2}k^2 = 28k^2.$$

Второе уравнение **сокращается на $k^2$** — и это главный ход задачи:
остаётся $n(n-1) = 56$, чистое уравнение на одно неизвестное.

**Отбор корней.** $n^2 - n - 56 = 0$ даёт $n = 8$ и $n = -7$. Второй корень
отбрасывают, потому что в условии сказано $n \in \mathbb{Z}^+$. За отбор
стоит балл, и «забыл посмотреть в условие» — самая частая его потеря
во всей теме.

**Симметрия против лишней работы.** Если в двух разных разложениях нужен
один и тот же $x^k$, номера $r$ в них разные, но $\binom{n}{r} = \binom{n}{n-r}$
позволяет считать с того конца, где степень ниже.

## Теория 4. Коэффициенты как последовательность

Небольшой, но регулярный вид: три коэффициента разложения связаны условием
из другой темы — арифметическое среднее, геометрическая прогрессия.
Работа делится надвое, и обе половины простые:

1. Формулой общего члена выписать три нужных коэффициента.
2. Записать условие связи и решить получившееся уравнение.

Условия, которые встречаются:

| Формулировка | Уравнение |
| --- | --- |
| третий член — среднее арифметическое второго и четвёртого | $2T_3 = T_2 + T_4$ |
| три коэффициента образуют геометрическую прогрессию | $\dfrac{T_3}{T_2} = \dfrac{T_4}{T_3}$ |

**Ловушка, за которую снимают последний балл.** Уравнения здесь получаются
из деления на $x$ или на $h$, и корень $0$ появляется почти всегда.
Он посторонний: при $x = 0$ разложение вырождается, при $h = 0$ прогрессии
нет. Markscheme про это пишет прямым текстом — «award final A0 for obtaining
$x = 0$». Посмотрите в условие: там почти наверняка написано $x \ne 0$
или $h > 0$.

## Теория 5. Дробный и отрицательный показатель

Вторая половина темы. Формула продолжается на любое действительное $m$,
но выглядит иначе — факториалов больше нет, потому что $\left(\frac12\right)!$
не существует:

$$(1+x)^m = 1 + mx + \frac{m(m-1)}{2!}x^2 + \frac{m(m-1)(m-2)}{3!}x^3 + \dots$$

Числитель — произведение $m$, $m-1$, $m-2$, …, знаменатель — факториал
номера. При целом положительном $m$ произведение рано или поздно обнуляется,
и ряд обрывается; при дробном или отрицательном $m$ он **бесконечен**,
и в ответе пишут столько членов, сколько просят.

**Два обязательных условия.**

*Форма $(1+\dots)^m$.* Формула работает только когда первое слагаемое —
единица. Выражение $\dfrac{1}{2-qx}$ сначала приводят к нужному виду:

$$\frac{1}{2-qx} = \frac{1}{2}\left(1 - \frac{q}{2}x\right)^{-1}.$$

Забыть вынести двойку — потерять всё разложение.

*Область сходимости.* Ряд имеет смысл только при $|x| < 1$. Если в скобке
стоит $ax$, условие превращается в $|ax| < 1$, то есть

$$|x| < \frac{1}{|a|}.$$

Это отдельный балл, и достаётся он одной строкой. В ноябре 2021 за ответ
«$|x| < 1$» стоял целый A1 — за одну строку, которую половина не пишет.

**Как считать без ошибок.** Коэффициенты берут по одному, не пытаясь
свернуть общую формулу. Для $(1+5x)^{1/2}$:

$$1,\qquad \tfrac12(5x),\qquad \frac{\frac12\left(-\frac12\right)}{2}(5x)^2,
\qquad \frac{\frac12\left(-\frac12\right)\left(-\frac32\right)}{6}(5x)^3.$$

Скобка $(5x)$ возводится в степень целиком — забыть возвести пятёрку
и есть самая частая ошибка приёма.

## Теория 6. Произведение рядов и приближение

**Перемножить и обрубить.** Если просят разложение произведения до $x^2$,
каждую скобку раскладывают до $x^2$, перемножают и **выбрасывают всё выше
$x^2$**. Считать члены, которые всё равно уйдут, — потерянное время:

$$(1+px)\left(1 - qx + q^2x^2\right)
= 1 + (p-q)x + \left(q^2-pq\right)x^2 + \underbrace{\dots}_{\text{в мусор}}$$

Полезная проверка: степень результата не выше требуемой, и слагаемых
ровно столько, сколько просили.

**Приближение числом.** Разложение — многочлен, в него можно подставить
число и получить приближение исходного выражения. Порядок:

1. Приравнять выражение под корнем к нужному числу и найти $x$.
2. Подставить этот $x$ в разложение.
3. Проверить, что $x$ лежит в области сходимости.

Пример: чтобы приблизить $\sqrt{1.2}$ через $(1+5x)^{1/2}$, решают
$1+5x = 1.2$, откуда $x = 0.04$.

**Откуда берётся точность.** Отброшенный хвост тем меньше, чем меньше $|x|$:
следующий член пропорционален $x^4$, и при $x = 0.04$ он в разы меньше,
чем при $x = 0.05$. Отсюда стандартный вопрос «почему это приближение хуже»
и стандартный ответ: **подставляемый $x$ больше**.

Крайний случай стоит различать отдельно. Если найденный $x$ вылезает
за $|x| < \frac{1}{|a|}$, приближение не «менее точное» — оно вообще
не имеет смысла, ряд расходится.

## Теория 7. Бином и комплексные числа

Стык с A5 и A6. Если в бином подставить $i$, разложение распадается
на действительную и мнимую части, и это даёт тригонометрические тождества.

$$(\cos\theta + i\sin\theta)^n
= \sum_{r=0}^{n}\binom{n}{r}\cos^{\,n-r}\theta\;\left(i\sin\theta\right)^{r}.$$

Степени $i$ ходят по кругу с периодом 4:

**Зачем это нужно.** По теореме Муавра то же самое равно
$\cos n\theta + i\sin n\theta$. Приравняв части, получают формулы кратного
угла — тот самый приём, которым в A6 доказывают $\sin 5\theta$. Здесь

## Решение 9 — геометрическая прогрессия

Разложение $(x+h)^8$ по убыванию степеней $x$:

$$x^8 + \binom81 hx^7 + \binom82 h^2x^6 + \binom83 h^3x^5 + \binom84 h^4x^4 + \dots$$

Значит

$$a = \binom81 h = 8h, \qquad b = \binom82 h^2 = 28h^2,
\qquad d = \binom84 h^4 = 70h^4.$$

Три числа образуют геометрическую прогрессию, когда отношения соседних
равны:

$$\frac{b}{a} = \frac{d}{b} \;\Longrightarrow\;
\frac{28h^2}{8h} = \frac{70h^4}{28h^2}
\;\Longrightarrow\; \frac{7h}{2} = \frac{5h^2}{2}.$$

Отсюда $7h = 5h^2$, то есть $h(5h-7) = 0$ и $h = \frac75 = 1{,}4$
(корень $h = 0$ отброшен, так как $h > 0$). $\blacksquare$


A1 за $h = 1{,}4$.

## Решение 12 — разность двух рядов

**(a)** Раскладываем каждую скобку до $x^2$.

$$(1+ax)^{-1/2} = 1 - \frac{a}{2}x + \frac{3a^2}{8}x^2 + \dots$$

$$(1-x)^{1/2} = 1 - \frac12 x - \frac18 x^2 + \dots$$

Вычитаем — единицы сокращаются:

$$(1+ax)^{-1/2} - (1-x)^{1/2}
= \frac{1-a}{2}x + \frac{3a^2+1}{8}x^2 + \dots$$

Приравниваем к $4bx + bx^2$:

$$\frac{1-a}{2} = 4b, \qquad \frac{3a^2+1}{8} = b.$$

Из первого $1-a = 8b$, из второго $3a^2+1 = 8b$. Значит

$$1 - a = 3a^2 + 1 \;\Longrightarrow\; 3a^2 + a = 0
\;\Longrightarrow\; a(3a+1) = 0.$$

Оба корня надо разобрать честно, потому что **оба удовлетворяют системе**.

При $a = 0$ выходит $b = \frac18$, и это не посторонний корень: левая часть
превращается в $1 - (1-x)^{1/2} = \frac{x}{2} + \frac{x^2}{8} + \dots$,
а правая при $b = \frac18$ равна $\frac{x}{2} + \frac{x^2}{8}$ — сходится
тождественно. Алгебра этот корень не отсекает.

Отсекает его условие $a \ne 0$: при $a = 0$ первая скобка вырождается
в единицу, и никакого разложения с показателем $-\frac12$ в задаче
не остаётся. Значит

$$a = -\frac13, \qquad b = \frac{1-a}{8} = \frac{1}{6}. \qquad \blacksquare$$

**(b)** Оба разложения требуют $|x| < 1$: для $(1-x)^{1/2}$ это прямо
условие $|x|<1$, для $(1+ax)^{-1/2}$ при $a = -\frac13$ — условие
$|x| < 3$. Пересечение даёт **$|x| < 1$**. $\blacksquare$

```python
my12_a = Rational(-1, 3)
my12_b = Rational(1, 6)
my12_r = 1
```

**Разметка.** M1 за попытку разложить с отрицательным дробным показателем,
A1 за первое разложение, A1 за второе, M1 за приравнивание коэффициентов,
M1 за решение системы, A1 за ответ. Пункт (b) — отдельный A1.

**Про пункт (b).** Один балл за одну строку — и его пропускают чаще, чем
любой другой в теме. Условие сходимости нужно писать всегда, когда
показатель нецелый, даже если про него не спросили отдельным пунктом:
разложение без области определения неполно.

**Как считать границу.** Общее правило: $(1+cx)^m$ сходится при $|cx|<1$,
то есть $|x| < \frac{1}{|c|}$. Когда скобок несколько, берут **самую
жёсткую** границу из всех. Здесь $(1-x)^{1/2}$ даёт $|x|<1$,
а $\left(1-\frac{x}{3}\right)^{-1/2}$ даёт $|x|<3$; побеждает первая.

**Третий раз одна и та же мысль.** В заданиях 8 и 9 лишний корень

Разница в том, что там корень был посторонним (появлялся от деления),
а тут он настоящий: системе он удовлетворяет. Тем важнее прочитать
ограничения — алгебра сама выбор не сделает.

## Решение 13 — произведение двух рядов

Знаменатель раскладываем как степень с показателем $-1$:

$$(1+qx)^{-1} = 1 - qx + q^2x^2 - \dots$$

Умножаем на числитель и держим только то, что не выше $x^2$:

$$(1+px)\left(1 - qx + q^2x^2\right)
= 1 - qx + q^2x^2 + px - pqx^2 + \underbrace{pq^2x^3}_{\text{выбрасываем}}$$

$$= 1 + (p-q)x + \left(q^2-pq\right)x^2 + \dots \qquad \blacksquare$$

```python
my13 = 1 + (p - q)*x + (q**2 - p*q)*x**2
```

**Разметка.** A1 за три верных члена разложения $(1+qx)^{-1}$, M1 за попытку
перемножить, A1 за ответ.

**Про экономию.** Раскладывать знаменатель дальше $x^2$ незачем: умножение
на $px$ поднимет степень, и всё лишнее всё равно уйдёт. Правило простое —
если нужен ответ до $x^k$, каждый сомножитель раскладывают тоже до $x^k$,
не дальше.

**Проверка себя.** Коэффициент при $x^2$ можно записать как $q(q-p)$ —
та же величина, и проверка это примет: сверяются значения, а не запись.

## Решение 16 — задание на таймере

**(c)** Первые три члена из задания 10:

$$(1+5x)^{1/2} = 1 + \frac52 x - \frac{25}{8}x^2 + \dots$$

$$\frac{1+px}{1+qx} = 1 + (p-q)x + \left(q^2-pq\right)x^2 + \dots$$

$$p - q = \frac52, \qquad q^2 - pq = -\frac{25}{8}.$$

$$q^2 - q\left(q+\frac52\right) = -\frac52 q = -\frac{25}{8}

**Где ошибаются в (d).** Читают $\frac{\sqrt5}{2}$ как $\sqrt{\frac52}$.
Разница принципиальная: $\frac{\sqrt5}{2} = \sqrt{1{,}25}$ даёт
$x = \frac1{20}$, а $\sqrt{2{,}5}$ потребовало бы $1+5x = 2{,}5$, то есть
$x = 0{,}3$. Но область сходимости здесь $|5x| < 1$, то есть $|x| < 0{,}2$ —
и $0{,}3$ в неё не попадает. Тогда ответ был бы не «менее точно»,
а «разложение вообще неприменимо». Полезно уметь различать эти два случая:
за пределами области сходимости разговор о точности теряет смысл.
