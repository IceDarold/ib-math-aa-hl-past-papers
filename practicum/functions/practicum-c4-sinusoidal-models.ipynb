## Теория 1. Четыре параметра и что делает каждый

$$H(t) = a\sin\bigl(b(t - c)\bigr) + d$$

| Параметр | Что означает | Как найти |
| --- | --- | --- |
| $a$ | амплитуда, половина размаха | $\dfrac{\max - \min}{2}$ |
| $d$ | средняя линия, вокруг которой колеблется | $\dfrac{\max + \min}{2}$ |
| $b$ | сжатие по горизонтали | $\dfrac{2\pi}{T}$, где $T$ — период |
| $c$ | сдвиг вправо по времени | из момента первого максимума |

Проверка, которая ловит половину ошибок: $a$ и $d$ **не зависят** от периода,
а $b$ и $c$ **не зависят** от высот. Если при поиске $b$ вам понадобился максимум,
где-то ошибка.

Ячейка ниже показывает, что делает каждый параметр. Меняйте числа и смотрите.

````python
grid = np.linspace(0, 24, 600)
model = lambda a, b, c, d: a*np.sin(b*(grid - c)) + d

fig, ax = plt.subplots(2, 2, figsize=(11, 5), sharex=True, sharey=True)
base = dict(a=2.3, b=np.pi/6, c=1.5, d=4.5)

for axis, (name, values, label) in zip(ax.flat, [
        ('a', [1.0, 2.3, 3.5], 'амплитуда: растягивает по вертикали'),
        ('d', [3.0, 4.5, 6.0], 'средняя линия: двигает вверх-вниз'),
        ('b', [np.pi/12, np.pi/6, np.pi/3], 'коэффициент b: сжимает по времени'),
        ('c', [0.0, 1.5, 3.0], 'сдвиг c: двигает вправо')]):
    for v in values:
        axis.plot(grid, model(**{**base, name: v}), lw=1.4, label=f'{name} = {v:.3g}')
    axis.set_title(label, fontsize=9)
    axis.legend(fontsize=7); axis.grid(alpha=.3)

fig.suptitle('H(t) = a·sin(b(t − c)) + d', fontsize=10)
plt.tight_layout(); plt.show()
````

## Теория 2. Амплитуда и средняя линия

Обе величины читаются из максимума и минимума, и обе — арифметика в одну строку:

$$a = \frac{\max - \min}{2}, \qquad d = \frac{\max + \min}{2}.$$

Смысл прозрачен на картинке: $d$ — это уровень, вокруг которого всё колеблется,
а $a$ — насколько далеко от него уходит график. Сумма даёт максимум, разность —
минимум, и это готовая проверка: $d + a$ обязано равняться максимуму.

Отсюда же второй способ, который markscheme принимает наравне с формулами:
составить систему $d + a = \max$, $d - a = \min$ и решить её. Способ длиннее,
но если формулы забылись, он выручает.

## Теория 3. Когда амплитуда отрицательна

Формула $a = \dfrac{\max - \min}{2}$ даёт **величину** амплитуды, всегда
положительную. Но в ответе $a$ иногда отрицательна, и это не описка.

Разберём, откуда берётся знак. Возьмём модель без сдвига: $H(t) = a\cos(bt) + d$.
При $t = 0$ косинус равен единице, поэтому $H(0) = a + d$:

- если $a > 0$, движение **начинается в максимуме**;
- если $a < 0$, движение **начинается в минимуме**.

Значит фраза «груз отпустили из нижней точки» или «кабинка стартует внизу»
— это указание на знак. Минус перед амплитудой переворачивает график
и есть способ сказать «начинаем снизу», не вводя сдвиг $c$.

Проверка занимает секунду: подставьте $t = 0$ и посмотрите, получился ли
максимум или минимум из условия.

## Теория 4. Фазовый сдвиг

$c$ отвечает на вопрос «когда» и находится из момента, в который случился
известный вам экстремум.

Ключ — где стоит максимум у неподвинутого синуса. У $\sin u$ максимум при
$u = \dfrac{\pi}{2}$, то есть через **четверть периода** после начала. Значит
у $a\sin\bigl(b(t - c)\bigr)$ с положительной $a$ максимум наступает при

$$b(t - c) = \frac{\pi}{2} \quad\Longleftrightarrow\quad t = c + \frac{T}{4}.$$

Отсюда правило в одну строку: **$c$ = момент первого максимума минус четверть
периода.**

Второй способ, который markscheme принимает наравне: подставить точку максимума
$(t_{\max}, H_{\max})$ в модель и решить получившееся уравнение относительно $c$.
Он надёжнее, когда вы не уверены, максимум перед вами или минимум, но требует
аккуратности с выбором наименьшего положительного корня — уравнение
$\sin(\cdot) = 1$ имеет их бесконечно много.

## Теория 7. Оценка через ±1

Приём, который заменяет вычисления рассуждением. Основан на единственном факте:

$$-1 \le \sin(\text{что угодно}) \le 1.$$

Аргумент не важен: сколь угодно сложное выражение под синусом ничего не меняет.
Поэтому если в функции есть слагаемое $\sin(\ldots)$, вы всегда знаете границы
его вклада, не решая ни одного уравнения.

**Разберём на настоящем вопросе.** Высоты двух растений моделируются как
$h_A(t) = \sin(2t + 6) + 9t + 27$ и $h_B(t) = 8t + 32$. Требуется доказать,
что при $t > 6$ растение A всегда выше.

Считать точки пересечения бессмысленно — «всегда» через численные корни
не доказывается. Вместо этого берут разность:

$$h_A(t) - h_B(t) = \sin(2t + 6) + t - 5.$$

При $t > 6$ имеем $t - 5 > 1$. А синус не бывает меньше $-1$. Значит вся
разность строго больше $1 - 1 = 0$, то есть $h_A > h_B$. Доказано, и ни одного
вычисления.

Форма рассуждения всегда одна: **отделить синус от остального, оценить синус
константой, показать, что остального хватает с запасом.**
