## Теория 1. Сколько корней ожидать

Прежде чем решать, полезно знать ответ на вопрос «сколько их». Это единственная
надёжная защита от главной ошибки темы.

На каждом обороте $2\pi$:

- $\sin x = k$ и $\cos x = k$ дают **два** корня при $|k| < 1$, один при $|k| = 1$,
  ни одного при $|k| > 1$;
- $\tan x = k$ даёт **один** корень на период $\pi$, то есть два на обороте $2\pi$.

Дальше умножайте на число оборотов в области. Если область $[0, 4\pi]$ — оборотов два,
и корней у $\sin x = 1/2$ будет четыре, а не два.

Опорный угол ищется обратной функцией **от модуля значения**, а знак и четверть
восстанавливаются по симметрии:

$$\sin x = k > 0 \Rightarrow x = \alpha,\ \pi - \alpha; \qquad
\cos x = k > 0 \Rightarrow x = \alpha,\ 2\pi - \alpha,$$

где $\alpha$ — острый угол. Для отрицательных значений четверти сдвигаются,
и здесь рисунок надёжнее памяти.

## Теория 3. Как получить квадратное уравнение

Две функции в одном уравнении сводятся к одной двумя тождествами.

**Пифагорово**, когда есть квадрат одной и первая степень другой:

$$\cos^2 x = 1 - \sin^2 x, \qquad \sin^2 x = 1 - \cos^2 x.$$

**Двойного угла**, когда есть и $2x$, и $x$. Форму выбирайте под ту функцию,
которая уже стоит рядом:

$$\cos 2x = 1 - 2\sin^2 x \quad (\text{рядом } \sin x), \qquad
\cos 2x = 2\cos^2 x - 1 \quad (\text{рядом } \cos x).$$

Выбор формы — единственное содержательное решение в этом приёме. Возьмёте не ту,
и в уравнении окажутся обе функции вместо одной.

Дальше получается квадратное уравнение относительно $\sin x$ или $\cos x$.
После его решения **обязательно проверьте оба корня на попадание в $[-1, 1]$**:
корень вроде $\sin x = 2$ или $\cos x = 5/4$ отбрасывается сразу, и за это
в markscheme стоит отдельный балл.

## Теория 4. Не делить, а раскладывать

Самая дорогая ошибка темы выглядит безобидно. В уравнении

$$\cos x = \sin 2x = 2\sin x\cos x$$

хочется поделить обе части на $\cos x$ и получить $\sin x = 1/2$. Но деление
законно только при $\cos x \ne 0$, а там, где $\cos x = 0$, обе части
уравнения равны нулю — то есть это тоже решения, и их только что выбросили.

Правильный ход всегда один: **перенести всё в одну часть и вынести множитель**.

$$\cos x - 2\sin x\cos x = 0 \;\Longrightarrow\; \cos x\,(1 - 2\sin x) = 0.$$

Теперь видно оба семейства корней: из $\cos x = 0$ и из $\sin x = 1/2$.

Правило: **сокращать на выражение с переменной нельзя никогда**. Если очень
хочется — вынесите его и решите отдельно. Проверка в этом ноутбуке ловит
такую потерю специально.

## Теория 5. Сведение к тангенсу

Если $\sin$ и $\cos$ входят в одинаковой степени, уравнение однородно, и его
можно поделить на $\cos^n x$ — получится уравнение относительно тангенса:

$$\cos^2 x - 3\sin^2 x = 0 \;\Big/\; \cos^2 x \;\Longrightarrow\; 1 - 3\tan^2 x = 0.$$

Здесь деление уже законно, но с оговоркой, которую надо произнести:
**проверьте отдельно, не является ли $\cos x = 0$ решением**. В примере выше
при $\cos x = 0$ получается $-3\sin^2 x = -3 \ne 0$, значит не является,
и делить можно.

Второй капкан — извлечение корня. Из $\tan^2 x = 1/3$ следует
$\tan x = \pm 1/\sqrt{3}$, то есть **две** серии решений. Половина потерянных
баллов в таких вопросах — это забытый минус.

## Теория 6. Отбор корней

Приём, который применяется поверх остальных. Четыре ситуации, где найти корни —
только половина работы.

**Дробь равна нулю.** Нужен нулевой числитель *при ненулевом знаменателе*.
Если корень обнуляет обе части, он не решение.

**Оговорка в условии.** Записи вроде $\theta \ne \pi/4$ ставят не случайно:
именно этот корень и получится, если решать невнимательно.

**Возведение в квадрат.** Порождает посторонние корни всегда. После него
подстановка в **исходное** уравнение обязательна — не в промежуточное.

**Обратные функции.** У $\arcsin$ и $\arccos$ аргумент обязан лежать в $[-1, 1]$.
В уравнении с $\arccos 3x$ это сразу сужает $x$ до $[-1/3, 1/3]$, и половина
кандидатов отпадает до всяких вычислений.

часть, построить график и считать точки пересечения с осью.

Ячейка ниже делает ровно то, что делает GDC.

````python
import matplotlib.pyplot as plt
import numpy as np

f = lambda t: 2*np.cos(t)**2 + 5*np.sin(t) - 4
grid = np.linspace(0, float(2*P), 600)

plt.figure(figsize=(8, 3))
plt.axhline(0, color='0.6', lw=1)
plt.plot(grid, f(grid))
plt.xlabel('x'); plt.ylabel('2cos²x + 5sin x − 4'); plt.title('то, что показывает GDC')
plt.tight_layout(); plt.show()

# корни тем же численным способом, каким их находит калькулятор
sign = np.sign(f(grid))
for i in np.where(np.diff(sign) != 0)[0]:
    lo, hi = grid[i], grid[i+1]
    for _ in range(60):
        mid = (lo + hi)/2
        lo, hi = (lo, mid) if f(lo)*f(mid) <= 0 else (mid, hi)
    r = (lo + hi)/2
    print(f'корень x ≈ {r:.4f}   это {r/np.pi:.4f}·pi')
````

## Решение 1 — область шире одного оборота

1. Касание оси: $6 + 6\cos x = 0$, то есть $\cos x = -1$.
2. Значение $|k| = 1$, поэтому на каждом обороте ровно **один** корень.
   Область $[0, 4\pi]$ — это два оборота, значит корня два.
3. $\cos x = -1$ при $x = \pi$ и, добавляя период $2\pi$, при $x = 3\pi$.

```python
my_roots = [P, 3*P]
```

## Решение 2 — градусы и составной аргумент

1. Пусть $u = 2x - 5^\circ$. Когда $x$ пробегает $[0^\circ, 180^\circ]$,
   $u$ пробегает $[-5^\circ, 355^\circ]$ — почти полный оборот.
2. $\tan u = 1$ при $u = 45^\circ$, период тангенса $180^\circ$,
   поэтому ещё $u = 225^\circ$. Оба попадают в область для $u$.
3. Возвращаемся: $2x = 50^\circ \Rightarrow x = 25^\circ$;
   $2x = 230^\circ \Rightarrow x = 115^\circ$.

```python
my_roots = [25, 115]
```

Если решать «для $x$, а потом делить», второй корень не появится.

## Решение 3 — наименьшее положительное

1. $\cos u = \dfrac{1}{\sqrt{2}}$ при $u = \pm\dfrac{\pi}{4} + 2k\pi$.
2. Здесь $u = \dfrac{x}{2} + \dfrac{\pi}{3}$, откуда
   $x = 2u - \dfrac{2\pi}{3} = \pm\dfrac{\pi}{2} - \dfrac{2\pi}{3} + 4k\pi$.
3. При $k = 0$ получаем $x = -\dfrac{\pi}{6}$ и $x = -\dfrac{7\pi}{6}$ — оба отрицательны.
4. Шагаем на период вперёд, $k = 1$: $x = \dfrac{23\pi}{6}$ и $x = \dfrac{17\pi}{6}$.
5. Наименьшее положительное — $\dfrac{17\pi}{6}$.

```python
my_x = 17*P/6
```

Период по $x$ здесь равен $4\pi$, а не $2\pi$: аргумент содержит $x/2$.
Кто шагает на $2\pi$, получает не тот ответ.

## Решение 4 — пифагорово тождество

1. Есть $\cos^2 x$ и первая степень $\sin x$, значит заменяем косинус:
   $2(1 - \sin^2 x) + 5\sin x = 4$.
2. Приводим: $-2\sin^2 x + 5\sin x - 2 = 0$, то есть $2\sin^2 x - 5\sin x + 2 = 0$.
3. Раскладываем: $(2\sin x - 1)(\sin x - 2) = 0$.
4. $\sin x = 2$ отбрасываем: синус не превосходит единицы. Остаётся $\sin x = \dfrac{1}{2}$.
5. На обороте $[0, 2\pi]$ это два корня: $x = \dfrac{\pi}{6}$ и $x = \dfrac{5\pi}{6}$.

```python
my_roots = [P/6, 5*P/6]
```

## Решение 5 — двойной угол

1. Рядом стоит $\sin x$, поэтому берём $\cos 2x = 1 - 2\sin^2 x$.
2. $1 - 2\sin^2 x = \sin x$, то есть $2\sin^2 x + \sin x - 1 = 0$.
3. $(2\sin x - 1)(\sin x + 1) = 0$, откуда $\sin x = \dfrac{1}{2}$ или $\sin x = -1$.
4. На $[-\pi, \pi]$: из $\sin x = \dfrac{1}{2}$ получаем $x = \dfrac{\pi}{6},\ \dfrac{5\pi}{6}$;
   из $\sin x = -1$ получаем $x = -\dfrac{\pi}{2}$.

```python
my_roots = [-P/2, P/6, 5*P/6]
```

Корень $-\dfrac{\pi}{2}$ — тот, который теряют: область захватывает
отрицательную половину, а привычка ищет корни в $[0, 2\pi]$.

## Решение 6 — невозможный корень

1. Рядом стоит $\cos\theta$, поэтому берём $\cos 2\theta = 2\cos^2\theta - 1$.
2. $2(2\cos^2\theta - 1) - 5\cos\theta + 2 = 0$, то есть $4\cos^2\theta - 5\cos\theta = 0$.
3. Выносим множитель: $\cos\theta\,(4\cos\theta - 5) = 0$.
4. $\cos\theta = \dfrac{5}{4} > 1$ невозможен. Остаётся $\cos\theta = 0$.
5. На $[\pi, 2\pi]$ это единственная точка $\theta = \dfrac{3\pi}{2}$.

```python
my_roots = [3*P/2]
```

Обратите внимание на шаг 3: свободного члена не осталось, и уравнение
раскладывается вынесением множителя, а не формулой корней.

## Решение 7 — разложение вместо деления

1. $\cos x = \sin 2x = 2\sin x\cos x$.
2. Переносим всё влево: $\cos x - 2\sin x\cos x = 0$.
3. Выносим: $\cos x\,(1 - 2\sin x) = 0$.
4. $\cos x = 0$ на $[0, \pi]$ даёт $x = \dfrac{\pi}{2}$.
5. $\sin x = \dfrac{1}{2}$ на $[0, \pi]$ даёт $x = \dfrac{\pi}{6},\ \dfrac{5\pi}{6}$.

```python
my_roots = [P/6, P/2, 5*P/6]
```

Если поделить на $\cos x$, останется $\sin x = 1/2$ и два корня вместо трёх.
Проверка в задании ловит именно это и говорит, около какой точки пропажа.

## Решение 8 — сведение к тангенсу

1. Проверяем, годится ли деление: при $\cos x = 0$ выражение равно
   $-3\sin^2 x = -3 \ne 0$, значит $\cos x = 0$ решением не является и делить можно.
2. Делим на $\cos^2 x$: $1 - 3\tan^2 x = 0$, откуда $\tan^2 x = \dfrac{1}{3}$.
3. Извлекаем корень с **двумя** знаками: $\tan x = \pm\dfrac{1}{\sqrt{3}}$.
4. На $[0, \pi]$: $\tan x = \dfrac{1}{\sqrt{3}}$ даёт $x = \dfrac{\pi}{6}$;
   $\tan x = -\dfrac{1}{\sqrt{3}}$ даёт $x = \dfrac{5\pi}{6}$.

```python
my_roots = [P/6, 5*P/6]
```

Шаг 1 стоит проговаривать вслух даже когда ответ очевиден: в вопросах,
где $\cos x = 0$ всё-таки является решением, за пропуск снимают балл.

## Решение 9 — дробь и оговорка

1. Дробь равна нулю, когда числитель ноль, а знаменатель нет.
2. Числитель: $2\sin^2 2\theta - 5\sin 2\theta - 3 = 0$. Обозначим $s = \sin 2\theta$:
   $2s^2 - 5s - 3 = 0$, то есть $(2s + 1)(s - 3) = 0$.
3. $s = 3$ невозможен. Остаётся $\sin 2\theta = -\dfrac{1}{2}$.
4. Знаменатель обращается в ноль при $\sin 2\theta = 1$ — это другое значение,
   так что найденные корни законны. Оговорка $\theta \ne \pi/4$ как раз и убирает
   точку, где знаменатель нулевой.
5. Область: $\theta \in [0, \pi]$ значит $2\theta \in [0, 2\pi]$ — полный оборот.
   $\sin u = -\dfrac{1}{2}$ даёт $u = \dfrac{7\pi}{6},\ \dfrac{11\pi}{6}$.
6. Возвращаемся: $\theta = \dfrac{7\pi}{12},\ \dfrac{11\pi}{12}$.

```python
my_roots = [7*P/12, 11*P/12]
```

## Решение 10 — посторонний корень

1. Область определения: нужен $|3x| \le 1$, то есть $x \in \left[-\dfrac{1}{3}, \dfrac{1}{3}\right]$.
   Это стоит выписать **до** решения.
2. Берём косинус от обеих частей. Пусть $A = \arccos x$, $B = \arccos 3x$:
   $$\cos(A + B) = \cos A\cos B - \sin A\sin B = \cos\frac{3\pi}{2} = 0.$$
3. Подставляем $\cos A = x$, $\cos B = 3x$, $\sin A = \sqrt{1 - x^2}$,
   $\sin B = \sqrt{1 - 9x^2}$:
   $$3x^2 = \sqrt{(1 - x^2)(1 - 9x^2)}.$$
4. Возводим в квадрат: $9x^4 = 1 - 10x^2 + 9x^4$, откуда $x^2 = \dfrac{1}{10}$
   и $x = \pm\dfrac{1}{\sqrt{10}}$. Оба попадают в область из шага 1.
5. **Проверка подстановкой.** При $x = \dfrac{1}{\sqrt{10}}$:
   $\arccos(0{,}316) + \arccos(0{,}949) \approx 1{,}249 + 0{,}322 = \dfrac{\pi}{2}$ — не подходит.
   При $x = -\dfrac{1}{\sqrt{10}}$:
   $\arccos(-0{,}316) + \arccos(-0{,}949) \approx 1{,}893 + 2{,}820 = \dfrac{3\pi}{2}$ — подходит.

$$\boxed{x = -\frac{1}{\sqrt{10}}}$$

```python
my_x = -1/sp.sqrt(10)
```

Почему посторонний корень возник: на шаге 3 левая часть $3x^2$ неотрицательна,
а возведение в квадрат стирает знак и пускает решение уравнения
$\cos(A+B) = 0$ с суммой $\pi/2$ вместо $3\pi/2$.

Отдельно: в разметке архива для этого вопроса записан ответ $x = \pm 1/2$.
Он неверен — при таком $x$ величина $\arccos 3x$ вообще не определена.

## Решение 12 — задание на таймере

4. Второй множитель: $\sin x = -\dfrac{1}{2}$.
   На $(0, 2\pi)$ это $x = \dfrac{7\pi}{6}$ и $x = \dfrac{11\pi}{6}$.
